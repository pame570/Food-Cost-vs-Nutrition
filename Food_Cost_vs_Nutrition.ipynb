## 4. METHODS
### 4.1 Dataset
My project uses three main datasets that each contribute a different piece of the overall analysis.
* **MenuStat (NYC DOHMH)**: This provides nutrient information for thousands of restaurant menu items, including calories, macronutrients, sodium, sugar, and food category labels. This dataset forms the basis for calculating nutrient density.

* **USDA Food-at-Home Dataset (F-MAP)**: This contains average retail food prices (USD per 100 grams) across detailed food groups. These prices allow me to evaluate the cost side of the "nutrient value per dollar" comparison.

* **USDA Table 8 (Dietary Guidelines)**: This provides the recommended nutrient density per 1000 calories for key dietary components such as fiber, added sugars, sodium, vegetables, and protein foods. These recommendations help me determine how actual food categories align with ideal nutritional standards.


Together, these datasets allow me to explore how nutrient quality and cost intersect across major food categories.

### 5.2 Nutrient Contributions by Food Group

Here I compute and plot average protein, fiber, and sugar per dollar for each food group. The heatmap reveals trade‑offs between beneficial nutrients (protein, fiber) and less desirable ones (sugar) in terms of how much of each nutrient is purchased with one dollar.


```python
heatmap_df = (
    final_df
    .infer_objects(copy=False)
    .replace([np.inf, -np.inf], np.nan)
    .dropna(subset=["Avg_Price"])
    .assign(
        Protein_per_dollar=lambda x: x["Protein_per_100kcal"] / x["Avg_Price"],
        Fiber_per_dollar=lambda x: x["Dietary_Fiber_per_100kcal"] / x["Avg_Price"],
        Sugar_per_dollar=lambda x: x["Sugar_per_100kcal"] / x["Avg_Price"],
    )
)

# Capping extreme nutrient-per-dollar values at the 99th percentile
for col in ["Protein_per_dollar", "Fiber_per_dollar", "Sugar_per_dollar"]:
    q99 = heatmap_df[col].quantile(0.99)
    heatmap_df[col] = heatmap_df[col].clip(upper=q99)

heatmap_df = (
    heatmap_df
    .groupby("Food_Group")[["Protein_per_dollar", "Fiber_per_dollar", "Sugar_per_dollar"]]
    .mean()
    .reset_index()
    .melt(id_vars="Food_Group", var_name="Nutrient", value_name="Value")
)

hv.HeatMap(
    heatmap_df,
    kdims=["Nutrient", "Food_Group"],
    vdims="Value",
).opts(
    width=800,
    height=450,
    colorbar=True,
    cmap="viridis",
    title="Average Nutrient Density per Dollar by Food Group and Nutrient",
    xlabel="Nutrient",
    ylabel="Food Group",
    tools=["hover"],
)
```

**Figure 4. Average nutrient density per dollar by food group and nutrient (values capped at the 99th percentile)**


### 5.3 Item-level NDD vs Price Scatterplot

This scatterplot shows each menu item’s Nutrient Density per Dollar (NDD) against its average price. Because prices come from F‑MAP food groups, every item in a group shares the same price, so the vertical spread within each group reflects differences in nutrient content. By plotting item‑level points, I can see whether higher‑priced items consistently deliver better protein‑plus‑fiber value, or whether some low‑ and mid‑priced items punch above their weight in NDD.


```python
item_scatter_df = (
    final_df
    .infer_objects(copy=False)
    .replace([np.inf, -np.inf], np.nan)
    .dropna(subset=["NDD", "Avg_Price", "Food_Group"])
)

# To cap very large NDD values
upper = item_scatter_df["NDD"].quantile(0.99)
item_scatter_df = item_scatter_df[item_scatter_df["NDD"] <= upper]

hv.Scatter(
    item_scatter_df,
    kdims="Avg_Price",
    vdims=["NDD", "Food_Group"]
).opts(
    color="Food_Group",
    size=4,
    alpha=0.5,
    width=800,
    height=450,
    title="Item-level Nutrient Density per Dollar vs Price",
    xlabel="Average Food-Group Price (USD per 100 g)",
    ylabel="Nutrient Density per Dollar (NDD)",
    legend_position="right",
    tools=["hover"],
)
```

**Figure 5. Item-level Nutrient Density per Dollar vs. food-group price**


### 5.4 Fiber Density Alongside USDA Table 8 Benchmarks

The Table 8 food‑group values are measured in cup‑ or ounce‑equivalents per 1,000 kcal, not grams of fiber, so this chart is a side‑by‑side view for context rather than a like‑for‑like comparison. A stricter version would compare every group against Table 8's single dietary fiber recommendation (see Limitations).


```python
table8_compare = build_table8_compare(final_df, table8_df)

bars_actual = hv.Bars(
    table8_compare,
    kdims="Food_Group_t8",
    vdims="actual_per_1000kcal",
).opts(
    title="Fiber Density vs. Table 8 Food-Group Recommendations (per 1,000 kcal)",
    xlabel="Food Group", ylabel="Per 1,000 kcal (fiber in g; Table 8 in cup/oz-eq)",
    width=900, height=450, xrotation=45, tools=["hover"]
)

points_rec = hv.Scatter(
    table8_compare,
    kdims="Food_Group_t8",
    vdims="rec_per_1000kcal",
    label="Table 8 food-group recommendation",
).opts( 
    color="red",
    size=8,
    marker="triangle",
    tools=["hover"],
)

(bars_actual * points_rec).opts(legend_position="right")
```

## 6. CONCLUSION

Across the visualizations, a clear pattern emerges: **foods that are least expensive are not necessarily the least nutritious, and in many cases offer the highest nutrient density per dollar**. Vegetables stand out as the strongest example of this relationship. They have the highest average nutrient density per dollar while also maintaining a relatively low average price, suggesting that nutrient‑rich diets do not inherently require higher food costs.

Grains and prepared foods occupy a middle ground. Grains provide moderate nutrient density at modest prices, making them a cost‑effective staple, while prepared foods tend to be more expensive without a proportional increase in nutrient density. Meats show the highest average prices but only moderate nutrient density per dollar, indicating diminishing nutritional returns relative to cost. Sugars and sweets perform worst across nearly all metrics, combining low nutrient density with relatively higher prices and emphasizing their role as being nutritionally inefficient choices.

The heatmap further clarifies these trends by showing how different food groups contribute specific nutrients per dollar. Vegetables and grains provide substantially more fiber and protein per dollar compared to meats and prepared foods, while beverages and sugars are dominated by sugar content with minimal nutritional benefit. This highlights not only differences in overall nutrient density, but also qualitative differences in nutrient composition across food groups.

At the item level, the scatterplot reveals substantial variability within food groups. Because each item carries its food group's average price, this view mainly shows how nutrient content varies within groups rather than item-level price effects. Items in lower‑priced groups, especially vegetables and grains, tend to reach high NDD scores, while items in higher‑priced groups cluster at lower values.

The Table 8 comparison adds context on fiber across food groups, but because the food‑group recommendations and fiber densities use different units, it should be read as descriptive rather than as a test of whether groups meet guidelines.

Taken together, these findings indicate that nutritional value and affordability are not mutually exclusive. Strategic selection of vegetables and whole grains can maximize nutrient intake per dollar, while heavy reliance on meats, prepared foods, and sugary products may raise food costs without delivering equivalent nutritional benefits. Within the limitations of my analysis (restaurant menu items paired with average retail prices), there is evidence that economic constraints need not be a barrier to healthier eating, and that nutrient‑dense diets can be achieved through informed food‑group choices.



## 7. LIMITATIONS
- **Group-level prices.** F‑MAP reports average unit prices per food group, so every menu item in a group receives the same price. Differences in NDD within a group come from nutrient content, not price.
- **Different price and nutrient bases.** Nutrients are measured per 100 kcal, while F‑MAP prices are per 100 grams of food. NDD is therefore a relative index for comparing groups, not an absolute "nutrients per dollar spent" figure.
- **Restaurant items vs. grocery prices.** MenuStat describes restaurant menu items, while F‑MAP describes food‑at‑home retail prices, so NDD reflects nutrient profiles of restaurant foods valued at grocery prices.
- **Manual category mapping.** MenuStat categories were mapped to broad food groups by hand (for example, Fried Potatoes and Salads to Vegetables, Sandwiches and Pizza to Grains). Different mapping choices could change the rankings.
- **Narrow nutrient index.** NDD uses protein and fiber only; it does not penalize sodium, sugar, or saturated fat or reward vitamins and minerals.
- **Table 8 units.** The Table 8 food‑group recommendations are in cup‑ or ounce‑equivalents per 1,000 kcal, not grams of fiber, so Figure 6 is descriptive only.
- **Outlier handling.** Infinite and undefined NDD values were removed and extreme values capped at the 99th percentile for plotting, which affects the scale of Figures 4 and 5.
